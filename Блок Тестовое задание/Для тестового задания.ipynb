```python
# Фильтруем пользователей, пришедших 1 ноября (Размер когорты)
cohort_day0 = df_audience[df_audience['date'] == '2026-11-01']['user_id'].unique()

# Смотрим, кто из них вернулся 2 ноября (День 1)
cohort_day1 = df_audience[df_audience['date'] == '2026-11-02']['user_id'].unique()

# Находим пересечение (вернувшиеся пользователи)
returned_users = np.intersect1d(cohort_day0, cohort_day1)

# Считаем процент удержания
retention_d1 = (len(returned_users) / len(cohort_day0)) * 100
print(f"3. Retention 1-го дня: {retention_d1:.1f}%") # Ожидаемый результат: 26.6% [cite: 14]
```

```python
# 5. Пользовательская конверсия в просмотр объявления (в пользователях)
total_users = df_audience['user_id'].nunique()
users_with_adverts = df_audience[df_audience['view_adverts'] > 0]['user_id'].nunique()

conversion_ad = (users_with_adverts / total_users) * 100
print(f"5. Конверсия в просмотр рекламы: {conversion_ad:.1f}%") # Ожидаемый результат: 46.3% [cite: 24]

# 6. Среднее количество просмотренных объявлений на пользователя
# Группируем по пользователю, считаем сумму его просмотров за месяц, затем берем среднее по всем пользователям
avg_adverts = df_audience.groupby('user_id')['view_adverts'].sum().mean()
print(f"6. Среднее количество объявлений на пользователя: {avg_adverts:.1f}") # Ожидаемый результат: 2.9 [cite: 26]
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Входные данные из задания [cite: 67, 68]
payments = np.array([1003, 1099])        # Количество успешных транзакций (A и B) [cite: 67, 68]
visitors = np.array([100047501, 100001055]) # Общее количество посетителей (A и B) [cite: 67, 68]

# Расчет конверсий в процентах [cite: 70]
cr_A = (payments[0] / visitors[0]) * 100
cr_B = (payments[1] / visitors[1]) * 100

print(f"Конверсия группы А: {cr_A:.5f}%") # ~0.00100% [cite: 71]
print(f"Конверсия группы B: {cr_B:.5f}%") # ~0.00110% [cite: 72]

# Относительный прирост группы B к группе A
relative_growth = ((cr_B - cr_A) / cr_A) * 100
print(f"Относительный рост конверсии в группе B: {relative_growth:.1f}%") # ~9.5% [cite: 73]

# Проверка на статистическую значимость (Z-тест)
z_stat, p_value_z = proportions_ztest(payments, visitors, alternative='two-sided')
print(f"p-value для теста долей: {p_value_z}") 

if p_value_z < 0.05:
    print("Результат статистически значим. Отклоняем нулевую гипотезу.") [cite: 74]
else:
    print("Различия не значимы. Нулевая гипотеза не отклоняется.")
```
